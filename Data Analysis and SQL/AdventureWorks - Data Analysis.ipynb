## **Graph Distribution of Values:**

```python
# FactInternetSales: Sales Amount & Order Quantity:
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('FactInternetSales Distribution', fontweight='bold')
sns.histplot(FactInternetSales['SalesAmount'], bins=50, kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Sales Amount Distribution')
sns.histplot(FactInternetSales['OrderQuantity'], bins=30, kde=True, ax=axes[1], color='lightgreen')
axes[1].set_title('Order Quantity Distribution')
plt.tight_layout()
plt.show()
```

---
* **Extreme Right Skew:** Sales Amount distribution shows a massive spike near 0 with a long tail up to ~3,578, confirming that most online transactions are low-value with only a few high-ticket orders.
* **Discrete Price Clusters:** Small secondary peaks around 500-750, 2,300-2,500, and 3,500 reflect distinct product price tiers rather than a continuous spread of values.
* **Single-Item Transactions:** The Order Quantity graph shows a single bar at exactly 1.0 spanning all 60,398 records, visually confirming that every row represents a one-unit line item.
* **No Multi-Unit Rows:** The complete absence of bars at 2 or higher proves that multi-item purchases are logged as separate transaction rows rather than aggregated quantities within a single row.
---

```python
# FactResellerSales: Sales Amount
plt.figure(figsize=(10, 5))
sns.histplot(FactResellerSales['SalesAmount'], bins=50, kde=True, color='teal')
plt.title('FactResellerSales - Sales Amount Distribution', fontweight='bold')
plt.show()
```

---
* **Extreme Right Skew:** Distribution spikes at over 32k records in the lowest sales bin and decays sharply, confirming that most reseller transactions are small while a few large bulk orders form the tail.
* **Long Tail of Bulk Orders:** Sparse bars extend past 15,000 toward the ~27,893 maximum, indicating occasional very large wholesale purchases that disproportionately drive total reseller revenue.
* **Secondary Order-Size Clusters:** Small bumps around 2,000 to 4,500 reflect discrete product price tiers and typical bulk order quantities rather than a smooth, continuous spread of values.
* **Clear B2B vs B2C Contrast:** Reseller tail stretches far beyond the internet sales maximum, visually confirming that B2B orders operate at a much larger transaction scale than consumer sales.
* **Analytical Implication:** The heavy skew means the mean overstates the typical transaction, so median-based reporting and outlier-aware visuals better represent actual reseller behavior.
